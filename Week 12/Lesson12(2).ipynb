```python
import cv2
import numpy as np

# Load image
img = cv2.imread("rice.png", 0)   # grayscale

# 1. Blur filter (remove noise)
blur = cv2.GaussianBlur(img, (5,5), 0)

# 2. Threshold (white objects become white, background black)
_, thresh = cv2.threshold(blur, 160, 255, cv2.THRESH_BINARY)

# 3. Morphology filter (clean image)
kernel = np.ones((3,3), np.uint8)
clean = cv2.morphologyEx(thresh, cv2.MORPH_OPEN, kernel)

# 4. Count objects
num_labels, labels, stats, centroids = cv2.connectedComponentsWithStats(clean)

# subtract 1 because label 0 = background
count = num_labels - 1

print("Number of objects:", count)

# Show result
cv2.imshow("Original", img)
cv2.imshow("Threshold", thresh)
cv2.imshow("Clean", clean)
cv2.waitKey(0)
cv2.destroyAllWindows()
```
