# Lesson 11

This notebook uses `finger.png` and applies binary morphology step by step.


```python
import numpy as np
import matplotlib.pyplot as plt
from PIL import Image

def binary_dilation(image, kernel):
    kh, kw = kernel.shape
    ph, pw = kh // 2, kw // 2
    padded = np.pad(image, ((ph, ph), (pw, pw)), mode='constant')
    windows = np.lib.stride_tricks.sliding_window_view(padded, kernel.shape)
    return np.any(windows & kernel, axis=(2, 3)).astype(np.uint8)

def binary_erosion(image, kernel):
    kh, kw = kernel.shape
    ph, pw = kh // 2, kw // 2
    padded = np.pad(image, ((ph, ph), (pw, pw)), mode='constant')
    windows = np.lib.stride_tricks.sliding_window_view(padded, kernel.shape)
    required = int(np.sum(kernel))
    return (np.sum(windows * kernel, axis=(2, 3)) == required).astype(np.uint8)

def binary_opening(image, kernel):
    return binary_dilation(binary_erosion(image, kernel), kernel)

def binary_closing(image, kernel):
    return binary_erosion(binary_dilation(image, kernel), kernel)

def show_binary(images, titles, cols=3, figsize=(12, 8)):
    rows = int(np.ceil(len(images) / cols))
    plt.figure(figsize=figsize)
    for i, (img, title) in enumerate(zip(images, titles), start=1):
        plt.subplot(rows, cols, i)
        plt.imshow(img, cmap='gray', vmin=0, vmax=1 if img.max() <= 1 else 255)
        plt.title(title)
        plt.axis('off')
    plt.tight_layout()
    plt.show()
```

## Task 1
Load `finger.png`, convert it to a binary image, and display the result.


```python
finger_img = Image.open('finger.png').convert('L')
finger_gray = np.array(finger_img)

# Threshold to create a binary image
threshold = 180
finger_binary = (finger_gray < threshold).astype(np.uint8)

print('finger image shape:', finger_gray.shape)
print('binary foreground pixels:', int(finger_binary.sum()))

show_binary(
    [finger_gray, finger_binary],
    ['finger.png Grayscale', 'Task 1: Binary Finger Image'],
    cols=2,
    figsize=(10, 4)
)
```

## Task 2
Use a **3 x 3** kernel and apply erosion, dilation, opening, and closing to the binary finger image.


```python
kernel_3 = np.ones((3, 3), dtype=np.uint8)

finger_eroded = binary_erosion(finger_binary, kernel_3)
finger_dilated = binary_dilation(finger_binary, kernel_3)
finger_opened = binary_opening(finger_binary, kernel_3)
finger_closed = binary_closing(finger_binary, kernel_3)

show_binary(
    [finger_binary, finger_eroded, finger_dilated, finger_opened, finger_closed],
    ['Original Binary', 'Erosion (3x3)', 'Dilation (3x3)', 'Opening (3x3)', 'Closing (3x3)'],
    cols=3,
    figsize=(12, 8)
)
```
